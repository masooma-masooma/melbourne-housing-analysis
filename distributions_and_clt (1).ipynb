## 1. Continuous Distribution

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

# Generate random numbers from a uniform distribution
data = uniform.rvs(size=1000, loc=0, scale=10)

# Plot the distribution
plt.hist(data, bins=30, density=True, alpha=0.6, color='g', edgecolor='black')
plt.title('Continuous Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## 2. Normal Distribution

```python

from scipy.stats import norm

# Generate data from a normal distribution
mu, sigma = 50, 10  # Mean and standard deviation
data = np.random.normal(mu, sigma, 1000)

# Plot the distribution
plt.hist(data, bins=30, density=True, alpha=0.6, color='b', edgecolor='black')

# Plot the theoretical normal distribution
x = np.linspace(min(data), max(data), 100)
plt.plot(x, norm.pdf(x, mu, sigma), 'r', linewidth=2)

plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```

## 3. Central Limit Theorem

```python

sample_means = [np.mean(np.random.normal(mu, sigma, 30)) for _ in range(1000)]

# Plot sample means distribution
plt.hist(sample_means, bins=30, density=True, alpha=0.6, color='c', edgecolor='black')
plt.title('Central Limit Theorem Demonstration')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.show()
```
